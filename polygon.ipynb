```python
ticker = "AAPL"

print("=" * 60)
print("Polygon.io 免费计划数据获取示例")
print("=" * 60)

# 免费计划支持的主要端点：
# 1. 获取公司信息
# 2. 获取股票拆分、分红等公司事件
# 3. 获取市场假日和交易日历

print("\n1. 获取股票基本信息")
print("-" * 60)
try:
    ticker_details = client.get_ticker_details(ticker=ticker)
    print(f"公司名: {ticker_details.name}")
    print(f"交易所: {ticker_details.primary_exchange}")
    print(f"行业: {ticker_details.sic_description if hasattr(ticker_details, 'sic_description') else 'N/A'}")
except Exception as e:
    print(f"获取失败: {e}")

print("\n2. 获取公司新闻")
print("-" * 60)
try:
    news = client.list_ticker_news(ticker=ticker, limit=5)
    for item in news:
        print(f"- {item.title}")
except Exception as e:
    print(f"获取失败: {e}")

print("\n3. 获取股票拆分信息")
print("-" * 60)
try:
    splits = client.list_splits(ticker=ticker)
    for split in splits:
        print(f"- 日期: {split.execution_date}, 比例: {split.split_from}/{split.split_to}")
except Exception as e:
    print(f"获取失败: {e}")

print("\n4. 获取分红信息")
print("-" * 60)
try:
    dividends = client.list_dividends(ticker=ticker)
    for div in dividends:
        print(f"- 日期: {div.ex_dividend_date}, 金额: ${div.dividend_per_share}")
except Exception as e:
    print(f"获取失败: {e}")

print("\n5. 获取股市日历（交易日和假日）")
print("-" * 60)
try:
    calendar = client.list_market_holidays()
    print(f"获取了 {len(list(calendar))} 个假日信息")
except Exception as e:
    print(f"获取失败: {e}")

print("\n" + "=" * 60)
print("提示：如需获取历史价格数据，请升级至付费计划")
print("=" * 60)
```
